```python
import os
import sys

import pymupdf as ppdf  # PyMuPDF — pip install pymupdf

# reuse path/file_name from the Netmonitor PDF export cell above - guarantees the PNG
# always matches the PDF that was just generated, no risk of rok/mesic drifting apart
path = fr'C:\Users\{user_name}\OneDrive - VLP\Dokumenty\01. Traffic and media\16. Netmonitor - obchod a kraje\{rok}'
file_name = f"M{mesic}-{rok}-Netmonitor_pro_obchod.pdf"
pdf_path = os.path.join(path, file_name)

if not os.path.isfile(pdf_path):
    sys.exit(f"PDF not found: {pdf_path}")

doc = ppdf.open(pdf_path)
zoom = 3  # ~216 DPI; raise for sharper output, lower for smaller files
matrix = ppdf.Matrix(zoom, zoom)

for i, page in enumerate(doc, start=1):
    pix = page.get_pixmap(matrix=matrix)
    suffix = "" if len(doc) == 1 else f"_p{i}"
    png_path = os.path.join(path, f"M{mesic}-{rok}-Netmonitor_pro_obchod{suffix}.png")
    pix.save(png_path)
    print(f"Saved: {png_path}")

doc.close()
```
